1. Basic Traversal & Manipulation


```python
# print All the Elements in a List

def print_list_elements(lst):
    for element in lst:
        print(element)

# Example usage

sample_list = [1, 2, 3, 4, 5]
print("Elements in the list are:")
print_list_elements(sample_list)
```

```python
#Find the maximum number in a list

def find_maximum(lst):
    if not lst:
        return None
    max_value = lst[0]
    for element in lst:
        if element > max_value:
            max_value = element
    return max_value
# Example usage
sample_list = [1, 2, 3,7, 4, 5]

max_value = find_maximum(sample_list)
print("The maximum value in the list is:", max_value)
```

```python
# sum of elements in lst

def sum(lst):
    if not lst:
        return 0
    total = 0
    for element in lst:
        total += element
    return total

sample_list = [1,2,3,4,5,6]

total = sum(sample_list)

print("the sum of elements in the list is:",total)
```

```python

# check number is exist in list
def check_number(lst,number):
    for element in lst:
        if element == number:
            return True

    return False

# Example usage

sample_list = [1,2,3,4,5]
number_to_check = 3

exist = check_number(sample_list,number_to_check)
if exist:
    print(f"the number {number_to_check} is exist in list")
else:
    print(f"the number {number_to_check} is not exist in list")
```

```python
#Reverse a list

def reverse_array(arr):
    left = 0
    right = len(arr) -1

    while left < right:
        arr[left],arr[right] = arr[right],arr[left]

        left += 1
        right -= 1

    return arr

arr = [1,7,8,3,4,5]

print(f"the reversed array is:{reverse_array(arr)}")

```

```python
	#Check if Array is Sorted

def is_sorted(arr):
    n = len(arr)
    for i in range(n - 1):
        if arr[i] >= arr[i+1]:
            return False
        
    return True

# Example Usage
arr = [ 1,2,3,4,5]
if is_sorted(arr):
    print("The Array is Sorted")
else:
    print("the array is not sorted")

```

```python
# 	Move All Zeros to End

def Move_Zeros_end(arr):
    non_zeros = []
    zeros = []

    for num in arr:
        if num!= 0:
            non_zeros.append(num)
        else:
            zeros.append(0)

    return non_zeros + zeros

arr = [1,3,0,8,0,5,0]

Move_Zeros_end(arr)
```

```python
	#Remove Duplicates from Sorted Array

def remove_duplicate(nums):
    if not nums:
        return 0
    
    insert_pos = 1

    for i in range(1,len(nums)):
        if nums[i] != nums[i-1]:
            nums[insert_pos] = nums[i]
            insert_pos += 1

    return insert_pos

# Example

nums = [1,2,3,4,4,15,2,3,1]
nums.sort()
k = remove_duplicate(nums)
print(k)
print(nums[:k])
```

 2. Two Pointer Technique
 

```python
#Pair with given sum in sorted array

def pair_with_sum(arr,target):
    left = 0
    right = len(arr) - 1

    while left < right :
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            print(f"Pair Found ({arr[left]} , {arr[right]}) = {current_sum}")
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    
    print("Pair Not found")
    return False

# Example input

arr = [1,2,3,4,5,6,7]
target = 10

pair_with_sum(arr,target)
```

```python
#Remove duplicates from sorted array

def remove_duplicates(nums):
    if not nums:
        return 0
    
    i = 0
    for j in range(1, len(nums)):
        if nums[j] != nums[i]:
            i += 1
            nums[i] = nums[j]
    
    return i + 1

#example

nums = [1,1,2,2,3]
length = remove_duplicates(nums)

print("length of unique array:",length)
print("unique values",nums[:length])

```

```python
# Move zeros to the end

def move_zeros(nums):
    pos = 0

    for i in range(len(nums)):
        if nums[i] != 0 :
            nums[pos] = nums[i]
            pos += 1

    while pos < len(nums):
        nums[pos] = 0
        pos += 1

    return nums

nums = [ 1,2,0,4,0,1]

move_zeros(nums)
```

```python
# Merge two sorted arrays

def merge_sorted_arrays(arr1,arr2):
    i,j = 0,0
    result = []

    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1

    while i < len(arr1):
        result.append(arr1[i])
        i += 1
    
    while j < len(arr2):
        result.append(arr2[j])
        j += 1

    return result

arr1 = [1,3,5,6,7]
arr2 = [2,4,6,8]

merge_sorted_arrays(arr1,arr2)

```
