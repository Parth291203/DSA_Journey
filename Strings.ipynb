**1. String Traversal & Basic Checks**

```python
# Leetcode 125 - Valid Palindrome

def ispalindrome(s:str):
    i ,j = 0,len(s) - 1

    while i < j:
        # Move i to next alnum:
        while i < j and not s[i].isalnum():
            i += 1
        # move j to the prev alnum:
        while i < j and not s[j].isalnum():
            j -= 1

        # Compare both

        if s[i].lower() != s[j].lower():
                return False
        
        i += 1
        j -= 1

    return True

si = "A man, a plan, a canal: Panama"

print(ispalindrome(si))
        

```

```python
# Leet code  28. Find the Index of the First Occurrence in a String
def compute_lps(pattern: str):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def strStr_kmp(haystack:str,needle : str):
    if needle == "":
        return 0
    n,m = len(haystack),len(needle)
    if m >n:
        return -1
    
    lps = compute_lps(needle)
    i= j = 0

    while i < n:
        if haystack[i] == needle[j]:
            i += 1
            j += 1
            if i == j :
                return i - j
        else:
            if j != 0:
                j = lps[j-1]
            else:
                i += 1

    return -1

haystack = "sadbutsad"
needle = "sad"

print(strStr_kmp(haystack,needle))
```

```python
# Leet code 58 - Length Of Last Word

class Solution:
    def lengthOfLastWord(self,s):

        i = len(s) - 1
        length = 0

        # 1. skip trailing spaces
        while i >= 0 and s[i] == ' ':
            i -= 1

        # 2. count the last word

        while i >= 0 and s[i] != ' ':
            

            i -= 1
            length += 1

        return length
    
obj = Solution()
    
input_str = "Hello World"

result = obj.lengthOfLastWord(input_str)

print(result)

    
```

**2. Frequency & Counting**

```python
# leetcode 387. First Unique Character in a String

def firstUniqChar(s):
    from collections import Counter
    freq = Counter(s)

    for i,chr in enumerate(s):
        if freq[chr] == 1:
            return i
    return -1
    
s = "loveleetcode"

print(firstUniqChar(s))
```

```python
# Leet code - 383 - Ransom Note

class Solution:
    def canConstruct(self,ransomNote,magazine):

        if len(ransomNote) > len(magazine):
            return False
        
        freq = [0] * 26

        # Build Supply

        for ch in magazine:
            freq[ord(ch) - ord('a')] += 1

        # Consume Demand

        for ch in ransomNote:
            idx = ord(ch) - ord('a')
            if freq[idx] == 0:
                return False
            freq[idx] -= 1

        return True
    
ransomNote = "aa"
magazine = "aab"

s = Solution()

result = s.canConstruct(ransomNote,magazine)

print(result)
```

```python
# Leet code 242 - Valid Anagrams

class Solution:
    def isAnagram(self,s:str,t:str):
        from collections import Counter

        return Counter(s) == Counter(t)
    
s = "anagram"
t = "nagaram"

obj = Solution()

result = obj.isAnagram(s,t)

print(result)
```

**3. String Reversal Patterns**

```python
# Leet code 344 - Reverse A String

class Solution:
    def reverseString(self,s :str):
        left,right = 0,len(s)-1

        while left < right:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1

s = ["h","e","l","l","o"]

obj = Solution()

result = obj.reverseString(s)

print(s)

```

```python
# Leet code - 557 - Reverse word in a string -III

class Solution:
    def reverseWords(self,s):
        arr = list(s)
        n = len(arr)
        i = 0

        while i < n:
            j = i
            while j < n and arr[j] != ' ':
                j += 1

            l,r = i,j-1
            while l < r:
                arr[l],arr[r] = arr[r],arr[l]
                l += 1
                r -= 1

            i = j + 1

        return ''.join(arr)
    
s = "Let's take LeetCode contest"

obj = Solution()

result = obj.reverseWords(s)

print(result)
```

```python
# leet code 345. Reverse Vowels of a String

class Solution:
    def reverseVowels(self,s):
        vowels = set("aeiouAEIOU")
        arr = list(s)
        i,j = 0,len(s)-1

        while i < j:
            while i < j and arr[i] not in vowels:
                i += 1

            while i < j and arr[j] not in vowels:
                j -= 1
            
            arr[i],arr[j] = arr[j],arr[i]
            i += 1
            j -= 1

        return "".join(arr)
    
s = "leetcode"

obj = Solution()

result = obj.reverseVowels(s)

print(result)
```

**4. String Transformations**

```python
# Leet code - 67. Add Binary (Binary String Addition)

class Solution:
    def AddBinary(self,a:str,b:str):
        i ,j = len(a) -1,len(b) -1
        carry = 0
        result = []

        while i >= 0 or j >= 0 or carry :
            bit_a = int(a[i]) if i >= 0 else 0
            bit_b = int(b[j]) if j >= 0 else 0

            total = bit_a + bit_b + carry

            result.append(str(total % 2))
            carry = total // 2

            i -= 1
            j -= 1

        return ''.join(reversed(result))
    

a = "1010"
b = "1011"

obj = Solution()

result = obj.AddBinary(a,b)

print(result)

```

```python
# Leet code 171 - Excel sheet column Number

def titleTONumber(columnTitle):
    if not columnTitle:
        return 0
    
    columnTitle = columnTitle.upper()
    result = 0

    for ch in columnTitle:
        # Convert Letters to 1..26

        val = ord(ch) - ord('A') + 1

        result = result * 26 + val

    return result

columnTitle = "AB"

print(titleTONumber(columnTitle))

```
