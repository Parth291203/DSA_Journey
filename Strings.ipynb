**1. String Traversal & Basic Checks**

```python
# Leetcode 125 - Valid Palindrome

def ispalindrome(s:str):
    i ,j = 0,len(s) - 1

    while i < j:
        # Move i to next alnum:
        while i < j and not s[i].isalnum():
            i += 1
        # move j to the prev alnum:
        while i < j and not s[j].isalnum():
            j -= 1

        # Compare both

        if s[i].lower() != s[j].lower():
                return False
        
        i += 1
        j -= 1

    return True

si = "A man, a plan, a canal: Panama"

print(ispalindrome(si))
        

```

```python
# Leet code  28. Find the Index of the First Occurrence in a String
def compute_lps(pattern: str):
    m = len(pattern)
    lps = [0] * m
    length = 0
    i = 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def strStr_kmp(haystack:str,needle : str):
    if needle == "":
        return 0
    n,m = len(haystack),len(needle)
    if m >n:
        return -1
    
    lps = compute_lps(needle)
    i= j = 0

    while i < n:
        if haystack[i] == needle[j]:
            i += 1
            j += 1
            if i == j :
                return i - j
        else:
            if j != 0:
                j = lps[j-1]
            else:
                i += 1

    return -1

haystack = "sadbutsad"
needle = "sad"

print(strStr_kmp(haystack,needle))
```

```python
# Leet code 58 - Length Of Last Word

class Solution:
    def lengthOfLastWord(self,s):

        i = len(s) - 1
        length = 0

        # 1. skip trailing spaces
        while i >= 0 and s[i] == ' ':
            i -= 1

        # 2. count the last word

        while i >= 0 and s[i] != ' ':
            

            i -= 1
            length += 1

        return length
    
obj = Solution()
    
input_str = "Hello World"

result = obj.lengthOfLastWord(input_str)

print(result)

    
```

**2. Frequency & Counting**

```python
# leetcode 387. First Unique Character in a String

def firstUniqChar(s):
    from collections import Counter
    freq = Counter(s)

    for i,chr in enumerate(s):
        if freq[chr] == 1:
            return i
    return -1
    
s = "loveleetcode"

print(firstUniqChar(s))
```

```python
# Leet code - 383 - Ransom Note

class Solution:
    def canConstruct(self,ransomNote,magazine):

        if len(ransomNote) > len(magazine):
            return False
        
        freq = [0] * 26

        # Build Supply

        for ch in magazine:
            freq[ord(ch) - ord('a')] += 1

        # Consume Demand

        for ch in ransomNote:
            idx = ord(ch) - ord('a')
            if freq[idx] == 0:
                return False
            freq[idx] -= 1

        return True
    
ransomNote = "aa"
magazine = "aab"

s = Solution()

result = s.canConstruct(ransomNote,magazine)

print(result)
```

```python
# Leet code 242 - Valid Anagrams

class Solution:
    def isAnagram(self,s:str,t:str):
        from collections import Counter

        return Counter(s) == Counter(t)
    
s = "anagram"
t = "nagaram"

obj = Solution()

result = obj.isAnagram(s,t)

print(result)
```

**3. String Reversal Patterns**

```python
# Leet code 344 - Reverse A String

class Solution:
    def reverseString(self,s :str):
        left,right = 0,len(s)-1

        while left < right:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1

s = ["h","e","l","l","o"]

obj = Solution()

result = obj.reverseString(s)

print(s)

```

```python
# Leet code - 557 - Reverse word in a string -III

class Solution:
    def reverseWords(self,s):
        arr = list(s)
        n = len(arr)
        i = 0

        while i < n:
            j = i
            while j < n and arr[j] != ' ':
                j += 1

            l,r = i,j-1
            while l < r:
                arr[l],arr[r] = arr[r],arr[l]
                l += 1
                r -= 1

            i = j + 1

        return ''.join(arr)
    
s = "Let's take LeetCode contest"

obj = Solution()

result = obj.reverseWords(s)

print(result)
```

```python
# leet code 345. Reverse Vowels of a String

class Solution:
    def reverseVowels(self,s):
        vowels = set("aeiouAEIOU")
        arr = list(s)
        i,j = 0,len(s)-1

        while i < j:
            while i < j and arr[i] not in vowels:
                i += 1

            while i < j and arr[j] not in vowels:
                j -= 1
            
            arr[i],arr[j] = arr[j],arr[i]
            i += 1
            j -= 1

        return "".join(arr)
    
s = "leetcode"

obj = Solution()

result = obj.reverseVowels(s)

print(result)
```

```python
# Leet code - 67. Add Binary (Binary String Addition)

class Solution:
    def AddBinary(self,a:str,b:str):
        i ,j = len(a) -1,len(b) -1
        carry = 0
        result = []

        while i >= 0 or j >= 0 or carry :
            bit_a = int(a[i]) if i >= 0 else 0
            bit_b = int(b[j]) if j >= 0 else 0

            total = bit_a + bit_b + carry

            result.append(str(total % 2))
            carry = total // 2

            i -= 1
            j -= 1

        return ''.join(reversed(result))
    

a = "1010"
b = "1011"

obj = Solution()

result = obj.AddBinary(a,b)

print(result)

```

```python
# Leet code 171 - Excel sheet column Number

def titleTONumber(columnTitle):
    if not columnTitle:
        return 0
    
    columnTitle = columnTitle.upper()
    result = 0

    for ch in columnTitle:
        # Convert Letters to 1..26

        val = ord(ch) - ord('A') + 1

        result = result * 26 + val

    return result

columnTitle = "AB"

print(titleTONumber(columnTitle))

```

```python
# Leet code - 12 - Integer to Roman

def intToRoman(num):
    values = [1000,900,500,400,100,90,50,40,10,9,5,4,1]
    symbols = ["M","CM","D","CD","C","XC","L","XL","X","IX","V","IV","I"]

    res = []

    for val,sym in zip(values,symbols):
        while num >= val:
            res.append(sym)

            num -= val

        if num == 0:
            break

    return "".join(res)

num = 1994


print(intToRoman(num))
```

**Advance Checks**

```python
# Leetcode - 392 - Is Subsequence

def isSubsequence(s,t):
    i,j = 0,0
    n,m = len(s),len(t)

    while i < n and j < m:
        if s[i] == t[j]:
            i += 1
        j += 1

    return i == n

s = "abc"
t = "ahbgdc"

print(isSubsequence(s,t))
```

```python
# Leet code - 796 - Rotate String

def rotateString(s,goal):
    if len(s) != len(goal):
        return False
    
    return goal in (s + s)
```

```python
# Leet code - 14 - Longest Common Prefix

class Solution:
    def longestCommonPrefix(self, strs):
        if not strs:
            return ""

        strs.sort()

        first = strs[0]
        last  = strs[-1]

        i =0
        while i < len(first) and i < len(last) and first[i] == last[i]:
            i += 1
        return first[:i]

solution = Solution()
print(solution.longestCommonPrefix(["flower","flow","flight"]))
```

🔹 Pattern 1: Sliding Window on Strings

```python
# Leet code - 3. Longest Substring Without Repeating Characters

def lengthOfLongestSubstring(s):
    last_seen = {}
    left  = 0
    max_len = 0

    for right,ch in enumerate(s):

        if ch in last_seen and last_seen[ch] >= left:
            left = last_seen[ch] + 1

        last_seen[ch] = right

        current_length = right - left + 1

        max_len = max(max_len,current_length)

    return max_len

s = "abcabcbb"

print(lengthOfLongestSubstring(s))


```

```python
# Leet code 567 - Permutation in string

def checkInclusion(s1,s2):
    from collections import Counter
    k = len(s1)
    if k > len(s2):
        return False
    
    need = Counter(s1)
    win = Counter()

    left = 0

    for right,ch in enumerate(s2):
        win[ch] += 1


        if right - left + 1 > k:
            left_char = s2[left]
            win[left_char] -= 1
            if win[left_char] == 0:
                del win[left_char]
            left += 1

        if right - left + 1 == k and win == need:
            return True
        
    return False
```
